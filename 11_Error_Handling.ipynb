### **What is Nested Error Handling in Python?**

```python
try:
    print("Outer try block")

    try:
        num = int(input("Enter a number: "))
        result = 10 / num
        print(result)

    except ValueError:
        print("Inner Error: Invalid input (not a number)")

    except ZeroDivisionError:
        print("Inner Error: Cannot divide by zero")

except Exception as e:
    print("Outer Error:", e)
```

```python
try:
    try:
        a = int(input("Enter a: "))
        b = int(input("Enter b: "))
        print(a / b)
    except ZeroDivisionError:
        print("Cannot divide by zero")
except:
    print("Something went wrong")
```
