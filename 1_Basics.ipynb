### **Escape sequences** 
* \ - for newline in editor
* \\\ - prints \
* \’ – prints ‘
* \” – prints “
* \n – prints a new line
* \t – prints a tab


```python
print("\\Hello\\world\\") # \Hello\world\
print('I\'m Python master')
print("Hello\"world\"")
print("Hello\nworld") # \n new line
print("hello\tworld") # \t = tab
```
