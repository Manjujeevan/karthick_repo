### **If, Elif Else**  *(Decision statement - True or False)*  
Syntax 
* **if \<condition\>:**
* **elif \<condition\>:**  
* **else:**

```python
a = 10

if a>100:   # : ->  Indentation 
    print(a,"is bigger then 100")

print("This line alwas work")

```

```python
a = 10

if a>100:  
    print(a,"is bigger then 100")

else:
    print(a,"is smaller then 100")

print("This line alwas work")
```

```python
a = 100

if a>100:  
    print(a,"is bigger then 100")

elif a == 100:
    print(a,"equal to 100")

else:
    print(a,"is smaller then 100")

print("This line alwas work")
```

### **If ,elif, else rules.**
* **if (Only once per code block(Mandtory))**
* **elif(as many time as per code block(optional))**
* **else(Only once per code black(optional))**  

```python
a = 4

if a>50:
    print("a is bigger")

elif a == 20:
    print("a is 20")

elif a == 30:
    print("a is 30")

elif a == 40:
    print("a is 40")

else:
    print("a is smaller")
    
```

### **Nested If and Else**

```python
a = -5

if a > 10:
    print("a is bigger")
    if a > 100:
        print("A IS BIGGEST")
    else:
        print("a is btwn 10 to 99")

elif a < 10:
    print("a is smaller")
    if a < 0:
        print("a is negative")
    else:
        print("a is btwn 0 to 9")



print("Code Over")
```
