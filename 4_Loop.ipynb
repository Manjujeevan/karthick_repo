### **Python Loop**  *(condition statements - True or Flase)*

**While Loop**

Syntax

* **while \<condition\>:**

```python
a = 1  # Initilization/Start 

while a<10:   # Condition/End 
    print(a)
    a += 1  #Incrementation/Jump

print("Loop Over")
```

```python
a = 1  

while a<=10: 
    print(a,end=" ")
    a += 1 

print("\nLoop Over")
```

```python
a = 100  

while a>=0: 
    print(a,end=" ")
    a -= 10

print("\nLoop Over")
```

### **break – stop the loop**


```python

# Infinite Loop
a = 1
while a>=0:
    print(a,end=' ')
    a += 1
```

```python

# Infinite Loop

a = 1
while a>=0: # True 
    if a == 20:
        
        break            #  loop stop
    print(a,end=' ')
    a += 1
```

### **continue – go back to the top of my loop**


```python
a = 0

while a<= 20:
    if a % 3 == 0:
        print(a,end=' ')
        a += 1
        continue
    
    a += 1
```

### **Nested While Loop**

```python
# Task
""" 
12345
12345
12345
12345
12345 
"""


a = 1

while a<=5:  # row
    b = 1
    while b<=5:   # column
        print(b,end="")
        b+=1
    a += 1
    print()



```

```python
a = 1

while a<=3:  # row
    b = 1
    while b<=3:   # column
        print(b,end="")
        b+=1
    a += 1
    print()

print("Loop Over")
```

```python
row = int(input("Enter a Row Number :"))
column = int(input("Enter a Collumn Number :"))

a = 1

while a<= row:
    b = 1
    while b<= column:
        print(b,end='')
        b += 1
    a += 1
    print()
```

### **Python For loop**  
Syntax

**for \<Variabul_name\> in \<str\\List\\Tuple\\Set\\Dict\>:**

```python
for x in "Hello world":  # Fetches Letter by letter
    print(x,end='')
print("\nLoop Over")
```

```python
mylist = [10,20.55,"hello",True,False]

print(mylist,type(mylist))

print("------------")

for x in mylist:               # Fetches Item by Item
    print(x,type(x))
```

```python
mylist = [10,20.55,"hello",True,False]

print(mylist,type(mylist))

print("------------")

for x in mylist:               # Fetches Item by Item
    print(x,end=', ')
```

### **range()** – *is a python standard function used for numbers*

```python
for x in range(20+1): #  end before
    print(x,end=' ')
```

```python
num = int(input("Enter a Number :"))

for x in  range(num+1):
    print(x,end=' ')
```

```python
for x in range(91,100+1): # start with end before
    print(x,end=' ')
```

```python
for x in range(0,100+1,10):  # Jump
    print(x,end=' ')
```

```python
# 100, 90,80,70,60,50,40,30,20,10

for x in range(100,0,-10):
    print(x,end=' ')
```

### **for with else**

```python
for x in "Hello world":
    print(x,end = "")

else:
    print("\nThe above loop complieted successfully")
```

* If Loop completes successfully, else will worke
* If loop is  unsuccessfully , else will not worke 

```python
for x in "Hello world":
    if x == ' ':
        break
    print(x,end = "")

else:
    print("\nThe above loop complieted successfully")
```

* Find if a number is divisible by 5 in the list

```python

mylist = [2,6,9,10,11,16,20,50,97]

for each in mylist:
    if each % 5 == 0:
        print(each,"Divisible By 5")
        break
else:
    print("Number divisible by 5 not found")        
```
