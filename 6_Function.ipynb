```python
def myfunc(x,y):
    print("X =",x)
    print("Y =",y)

    print("The sum of : ", x+y)

myfunc(100,200)  # arguments: 100,200
print()

myfunc(30,20)
print()
myfunc(20,70)
```

### **Arbitrary Arguments**
* *args - argument stores in tuple
* **kwargs - keyworded argument stores in dictionary


```python

def myfunc(*x):  # *args - argument stores in tuple
    print(x,type(x))


myfunc()
myfunc(10)
myfunc("Hai")
myfunc(True,50.55)
```

```python
def myfunc(**x):   # **kwargs - keyworded argument stores in dictionary
    return x

ans1 = myfunc(a = 10)
print(ans1)

ans2 = myfunc(Name = "Manjunathan", Age = 40, City = "Chennai")
print(ans2)
```

### **Function overloading**

```python
def myfunc(x):
    return x

def myfunc(x,y):
    return x+y


def myfunc(x,y,z):
    return x+y+z

ans = myfunc(30,40,30)
print(ans)
```
