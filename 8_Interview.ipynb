```python
# Interview   check palindrome

# 1 Find weather the String is check palindrome

# 1.  Input: racecar
#     Output: True
# 2.  Input: MalaYa lAm
#     Output: True
# 3   Input: Sir, I demand, I am a maid named Iris
#     Output: True
# 4   Input : Red roses run no risk. Sir, on Nurse's Order.
```

```python
mystr = 'racecar'
revstr = mystr[::-1]

print(mystr)
print(revstr)


if mystr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = 'MalaYa lAm'
newstr = mystr.lower().replace(' ','')
revstr = newstr[::-1]

print(mystr)
print(newstr)
print(revstr)


if newstr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = 'Sir, I demand, I am a maid named Iris'
newstr = mystr.lower().replace(' ','').replace(',','')
revstr = newstr[::-1]

print(mystr)
print(newstr)
print(revstr)


if newstr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = "Red roses run no risk. Sir, on Nurse's Order."

newstr = ''

for each in mystr:
    if each.isalpha():
        newstr = newstr+each.lower()   
print(newstr)


if newstr == newstr[::-1]:
    print(True)
else:
    print(False)
```

```python
def check_palindrome(mystr):
    newstr = ''

    for each in mystr:
        if each.isalpha():
            newstr = newstr+each.lower()   
    
    if newstr == newstr[::-1]:
        return True
    else:
        return False


ans = (check_palindrome("MalaYa lAm"))
print(ans)

print(check_palindrome("racecar"))
print(check_palindrome("Sir, I demand, I am a maid named Iris"))
print(check_palindrome("Red roses run no risk. Sir, on Nurse's Order."))
```
