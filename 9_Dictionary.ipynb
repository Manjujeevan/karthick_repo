### **Dictionary Functions**

```python
mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}


print(mydict,type(mydict))
print(mydict["Name"])
print(mydict["Age"])

mydict["City"]="Mumbai"
mydict["Cours"] = "Python"
print(mydict)
```

```python
# keys()

mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}


print(mydict)
print(list(mydict.keys()))
print()

for each in mydict.keys():
    print(each)
```

```python
# values()

mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}


print(mydict)
ans = mydict.values()

print(ans,type(ans))
print()
for each in mydict.values():
    print(each,type(each))
```

```python
# items()

mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}
print(mydict)

print(list(mydict.items()))
print()

for each in mydict.items():
    print(each)

```

```python
# items()

mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}
print(mydict)

print(list(mydict.items()))
print()

for each in mydict.items():
    print(f"{each[0]}\t{each[1]}")
```

```python
# pop()

# list = pop(),pop(index)

x = [10,30.55,"hello",10,30.55,"hai"]
print(x)
x.remove(30.55)
x.remove(30.55)

print(x)
```

```python
# set = pop()

x = {10,30.55,"hello",10,30.55,"hai"}

print(x)
x.pop()
print(x)
```

```python
# pop(key)

mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}
print(mydict)
mydict.pop("City")
print(mydict)
```

```python
# popitem()

mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}
print(mydict)
mydict.popitem()
mydict.popitem()
print(mydict)
```

```python
# clear()


mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}

print(mydict)

mydict.clear()
print(mydict)
```

```python
# update()


mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}

print(mydict)

newdict = {"Name":"Honey","Age":35,"City":"Kovai","Course" : "Python"}

mydict.update(newdict)

print(mydict)
```

```python
# get()

mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}

print(mydict)
print(mydict[500])
print(mydict[600])
print("Hello world")
```

```python
mydict ={
    "Name":"Manjunathan",
     "Age": 40,
    "City":"Kanchi",
       500:50.55,      
}

print(mydict)
print(mydict.get(500))
print(mydict.get(600))
print("Hello world")
```

```python
# copy()

x = {'A':100,'B':200,'C':300}
y = x

x['D']=400


print(x,type(x),id(x))
print(y,type(y),id(y))
```

```python

x = {'A':100,'B':200,'C':300}
y = x.copy()

x['D']=400
y['E']=500


print(x,type(x),id(x))
print(y,type(y),id(y))
```

### **Nested Dictionary**

```python
mydict = {
    "Name":"Jeevanathan",
    "City":["Chennai","Kovai","Mumbai"],
    "Year":(2000,2010,2020)
}

print(mydict,type(mydict))
print(mydict.keys())
print(mydict.get("Name"))
print(mydict.get("City"))
print(mydict.get("City")[0])
print(mydict.get("City")[1])
print(mydict.get("City")[2])
print('------------------------------------------')

print(mydict.get('Year'))
print(mydict.get('Year')[0])
print(mydict.get('Year')[1])
print(mydict.get('Year')[2])

print('------------------------------------------')

print(mydict.get("City")[2][3:])
```

```python

x = {
    "menu": {
        "id": "file",
        "value": "File",
        "popup": {
            "menuitem": [
                {"value": "New", "onclick": "CreateNewDoc()"},
                {"value": "Open", "onclick": "OpenDoc()"},
                {"value": "Close", "onclick": "CloseDoc()"}
            ]
        }
    }
}


print(x,type(x))
print(x.keys())
print(x.get('menu'))
print(x.get('menu')['popup'])
print(x.get('menu')['popup']['menuitem'])
print('----------------------------------------')
print(x.get('menu')['popup']['menuitem'][0])
print(x.get('menu')['popup']['menuitem'][1])
print(x.get('menu')['popup']['menuitem'][2])
print('----------------------------------------')
print(x.get('menu')['popup']['menuitem'][0]['onclick'])
print(x.get('menu')['popup']['menuitem'][1]['value'])
print(x.get('menu')['popup']['menuitem'][2]['onclick'])
```
